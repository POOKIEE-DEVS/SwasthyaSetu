```python
import os

from google.colab import userdata
from huggingface_hub import auth_check, whoami
from huggingface_hub.errors import GatedRepoError

MODEL_ID = "google/medgemma-1.5-4b-it"
os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
os.environ["GRADIO_SHARE"] = "1"

# Fails fast, with a clear reason, before the long model download.
print("Token belongs to Hugging Face account:", whoami()["name"])
try:
    auth_check(MODEL_ID)
    print("Access to", MODEL_ID, "OK")
except GatedRepoError:
    raise SystemExit(
        f"No access to {MODEL_ID}. Signed in as the account above, open "
        f"https://huggingface.co/{MODEL_ID} and accept the terms. If you use a "
        "fine-grained token, tick 'Read access to contents of all public gated "
        "repos you can access', or use a classic Read token."
    ) from None
```
